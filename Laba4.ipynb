```python
import sys
%matplotlib inline
import matplotlib.pyplot as plt
import cv2 as cv
import numpy as np

# Изменим стандартный размер графиков matplotlib
plt.rcParams["figure.figsize"] = [6, 4]

image = cv.imread('./text.jpg')
grayImage = cv.cvtColor(image, cv.COLOR_RGB2GRAY)
```

```python
plt.figure(figsize=(10, 6))
plt.hist(grayImage.reshape(-1), 256, [0, 256])
plt.title('Гистограмма яркости изображения')
plt.show()
```

```python
threshold = 115
ret1, thresh1 = cv.threshold(grayImage, threshold, 255, cv.THRESH_BINARY)

cv.imwrite('./text.jpg', thresh1)

plt.figure(figsize=(32, 2))
plt.imshow(thresh1, cmap='gray')
plt.title('Бинаризация полутоновых изображений', fontsize=14)
plt.xticks([]), plt.yticks([])
```
