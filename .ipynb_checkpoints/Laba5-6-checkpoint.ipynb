# Лабораторная работа 5-6
Выделить фрукты на изображении

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

image = cv.imread('oranges.jpg')
image_rgb = cv.cvtColor(image, cv.COLOR_BGR2RGB)

gs = plt.GridSpec(1, 2)
```

```python
plt.figure()
plt.title("1. Исходное изображение")
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```

Преобразование в HSV

```python
image_hsv = cv.cvtColor(image_rgb, cv.COLOR_RGB2HSV)

plt.figure()
plt.title("Изображение в HSV")
plt.imshow(image_hsv)
plt.axis("off")
plt.show()
```

Маска апельсинов

```python
lower_orange = np.array([5, 80, 80])
upper_orange = np.array([25, 255, 255])

mask_fruits = cv.inRange(image_hsv, lower_orange, upper_orange)

image_with_mask = cv.bitwise_and(image_rgb, image_rgb, mask=mask_fruits)

plt.figure(figsize=(10, 6))
plt.subplot(gs[0])
plt.title("Маска апельсинов")
plt.imshow(mask_fruits, cmap="gray")
plt.axis("off")

plt.subplot(gs[1])
plt.title("Апельсины с применённой маской")
plt.imshow(image_with_mask)
plt.axis("off")
plt.show()
```

```python
contours, _ = cv.findContours(mask_fruits, cv.RETR_EXTERNAL, cv.CHAIN_APPROX_SIMPLE)

output = image_rgb.copy()
for cnt in contours:
    cv.drawContours(output, [cnt], -1, (0, 255, 0), 3)

plt.figure()
plt.title("Выделенные фрукты (контуры)")
plt.imshow(output)
plt.axis("off")
plt.show()
```

Маска белого фона

```python
lower_white = np.array([0, 0, 200])
upper_white = np.array([180, 50, 255])

mask_white = cv.inRange(image_hsv, lower_white, upper_white)

plt.figure()
plt.title("Маска белого фона")
plt.imshow(mask_white, cmap="gray")
plt.axis("off")
plt.show()
```



Итоговая маска

```python
final_mask = cv.bitwise_and(mask_fruits, cv.bitwise_not(mask_white))

kernel = np.ones((5, 5), np.uint8)
final_mask = cv.morphologyEx(final_mask, cv.MORPH_CLOSE, kernel)
final_mask = cv.GaussianBlur(final_mask, (7, 7), 0)

plt.figure()
plt.title("Итоговая маска после очистки")
plt.imshow(final_mask, cmap="gray")
plt.axis("off")
plt.show()
```

Применение маски к изображению

```python
segmented = cv.bitwise_and(image_rgb, image_rgb, mask=final_mask)

plt.figure()
plt.title("Апельсины с применённой маской")
plt.imshow(segmented)
plt.axis("off")
plt.show()
```

Отображение контуров

```python
contours, _ = cv.findContours(final_mask, cv.RETR_EXTERNAL, cv.CHAIN_APPROX_SIMPLE)

output = image_rgb.copy()
for cnt in contours:
    cv.drawContours(output, [cnt], -1, (0, 255, 0), 3)

plt.figure()
plt.title("Выделенные фрукты (контуры)")
plt.imshow(output)
plt.axis("off")
plt.show()
```
